This week we're exploring data about links between cities! Twinned towns (also known as sister cities) are a form of legal or social agreement between two geographically and politically distinct localities for the purpose of promoting cultural and commercial ties. This dataset looks at links specifically between cities, i.e. it does not include towns, villages or other geographic entities..

Wikipedia states:

While there are early examples of international links between municipalities akin to what are now known as sister cities dating back to the 9th century, the modern concept was first established and adopted worldwide during World War II.

Is every country connected through a chain of twin city links?

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import geopandas as gp
```

```python
cities = pd.read_csv('https://raw.githubusercontent.com/rfordatascience/tidytuesday/main/data/2026/2026-05-12/cities.csv')
links = pd.read_csv('https://raw.githubusercontent.com/rfordatascience/tidytuesday/main/data/2026/2026-05-12/links.csv')

world_map_url = "https://naciscdn.org/naturalearth/110m/cultural/ne_110m_admin_0_countries.zip"
world = gp.read_file(world_map_url)

world = world.loc[:, ["SOVEREIGNT", "geometry"]].query("SOVEREIGNT != 'Antarctica'")
```

```python
cities.info()
```

```python
links.info()
```

```python
df = pd.merge(cities, links, how = "left", left_on = "id", right_on="source")
```

```python
world.head()
```

```python
df = gp.GeoDataFrame(df)
```

```python
# Group and count
counts_df = df.groupby('name').size().reset_index(name='occurrences')
```

```python
counts_df
```

```python
# Get unique locations for each name
unique_coords = df.drop_duplicates(subset=['name'])[['name', 'lat', 'lng']]

# Merge the bins onto the coordinates
map_data = unique_coords.merge(counts_df, on='name')
```

```python
map_data
```

```python
fig, ax = plt.subplots(1, 1, figsize=(16, 8))
world.plot(ax=ax, color='lightgrey', edgecolor='white', linewidth=0.5)

# 1. Use the raw occurrences directly for size (scaled by 5 or 10)
# We use a single scatter call for much better performance
scatter = ax.scatter(
    map_data['lng'], 
    map_data['lat'], 
    s=map_data['occurrences'] * 10, # Adjust multiplier as needed
    c=map_data['occurrences'],      # Optional: Color them by count too!
    cmap='viridis', 
    alpha=0.6,
    edgecolors='white',
    linewidth=0.5
)

# 2. Create a manual legend for specific values (so it's not cluttered)
# We pick a few representative counts to show in the legend
for count in [1, 10, 25, 50]:
    ax.scatter([], [], s=count * 10, c='gray', alpha=0.6, label=f'{count} Cities')

ax.legend(title="Twin Cities", loc='lower left', labelspacing=1.5, borderpad=1)
ax.set_title("Global Distribution of Twin Cities (Shared Names)")
ax.axis('Off')
plt.show()
```
